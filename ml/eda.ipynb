```python
# Calculating min and max values of count of length of text, words count for each value of Ticket Description column
count_of_text_length = df['Ticket Description'].str.len()
count_of_words = df['Ticket Description'].str.split().str.len()

# Minimum Values
min_of_count_text_length = count_of_text_length.min();
min_of_count_words = count_of_words.min();
print("min_of_count_text_length", min_of_count_text_length)
print("min_of_count_words", min_of_count_words)

# Maximum values
max_of_count_text_length = count_of_text_length.max();
max_of_count_words = count_of_words.max();
print("max_of_count_text_length", max_of_count_text_length)
print("max_of_count_words", max_of_count_words)
```

```python
# Calculating IQR for checking outliers in Text length

count_of_text_length = df['Ticket Description'].str.len()
count_of_words = df['Ticket Description'].str.split().str.len()


Q1 = count_of_text_length.quantile(0.25)
Q3 = count_of_text_length.quantile(0.75)

IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

outliers = (count_of_text_length < lower_fence) | (count_of_text_length > upper_fence)
print("Lower fence:", lower_fence, "Upper fence:", upper_fence)
print("Number of outliers:", outliers.sum())
print(count_of_text_length[outliers].describe())
```

```python
# Checking outliers for words count 
count_of_words = df['Ticket Description'].str.split().str.len()


Q1 = count_of_words.quantile(0.25)
Q3 = count_of_words.quantile(0.75)

IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

outliers = (count_of_words < lower_fence) | (count_of_words > upper_fence)
print("Lower fence:", lower_fence, "Upper fence:", upper_fence)
print("Number of outliers:", outliers.sum())
print(count_of_words[outliers].describe())
```

# (TIC - 15 ) Check missing values and data quality issues

```python
# Checking NULL values & Empty Values count sum in dataset columns
print("Column Name          " + " | " + "Total values count"  + " | " + "NULL values count" +  " | " + "Empty values count")
for column in df:
    total_value_count = df[column].size
    null_value_count = df[column].isna().sum()
    empty_value_count = (df[column] == '').sum()
    print(column + "        " +  " | " + str(total_value_count) + " | " + str(null_value_count) + " | " + str(empty_value_count))
```

```python
# Checking duplicated rows in dataset
duplicated = df.duplicated()
sum = 0
#print(duplicated)
if(duplicated.any()):
    sum += 1;

print("Duplicated rows:", sum)
```

```python
# Checking correlation of Ticket Status with all 4 columns with NULL values
# Resolution , Time to Resolution, Customer Satisfaction Rating, First Response Time
df.groupby('Ticket Status')[['Resolution', 'Time to Resolution', 'Customer Satisfaction Rating']].apply(lambda x: x.isnull().sum())
```
